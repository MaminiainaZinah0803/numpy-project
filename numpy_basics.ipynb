```python
# 1. Import the NumPy library and give it a short nickname 'np'
import numpy as np

# 2. Create a simple NumPy array (this is NumPy's version of a list)
my_array = np.array([10, 20, 30, 40, 50])

# 3. Print the array and its type to see the difference from a normal list
print("My Array:", my_array)
print("Type:", type(my_array))
```

```python
# 1. Vectorized Math: Add 5 to EVERY element at once!
print("Original array:", my_array)
print("Add 5 to all:", my_array + 5)

# 2. Multiply every element by 2
print("Multiply by 2:", my_array * 2)

# 3. NumPy Statistics (Very useful for data!)
print("\n--- Statistics ---")
print("Mean (Average):", np.mean(my_array))
print("Maximum value:", np.max(my_array))
print("Minimum value:", np.min(my_array))
print("Sum of all elements:", np.sum(my_array))
```
